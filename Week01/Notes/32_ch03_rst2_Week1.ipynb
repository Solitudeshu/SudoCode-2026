## Finding Word Stems

Pull out the stem of a word (tách gốc từ):

```python
def stem(word):
  for suffix in ["ing", "ly", "ed", "ious", "ies", "ive", "es", "s", "ment"]:
    if word.endswith(suffix):
      return word[:-len(suffix)]
  return word

w = "processing"
print(stem(w))
```

Regex stemmer

```python
re.findall(r'^.*(ing|ly|ed|ious|ies|ive|es|s|ment)$', "processing")
```

```python
re.findall(r'^.*(?:ing|ly|ed|ious|ies|ive|es|s|ment)$', "processing")
```

(...) là capturing group.

(?:...) = non-capturing group.

Cả hai đều group pattern, nhưng ?: nói:
Chỉ group để điều khiển regex; đừng lấy phần này ra làm output capture.

Tách gốc và hậu tố:

```python
re.findall(r'^(.*)(ing|ly|ed|ious|ies|ive|es|s|ment)$', "processing")
```

```python
# greedy
print(re.findall(r'^(.*)(ing|ly|ed|ious|ies|ive|es|s|ment)$', "processes"))

# non-greedy
print(re.findall(r'^(.*?)(ing|ly|ed|ious|ies|ive|es|s|ment)$', "processes"))

# ?$ suffix có thể có hoặc không
print(re.findall(r'^(.*?)(ing|ly|ed|ious|ies|ive|es|s|ment)?$', "language"))
```

Tạo index cho một văn bản dựa trên stem của từ, sau đó cho phép tìm tất cả vị trí xuất hiện của một từ và in phần văn bản xung quanh nó (concordance).

```python
class IndexedText(object):

    def __init__(self, stemmer, text):
        self._text = text
        self._stemmer = stemmer
        self._index = nltk.Index((self._stem(word), i)
                                  for (i, word) in enumerate(text))

    def concordance(self, word, width=40):
        key = self._stem(word)
        wc = int(width/4)                # words of context
        for i in self._index[key]:
            lcontext = ' '.join(self._text[i-wc:i])
            rcontext = ' '.join(self._text[i:i+wc])
            ldisplay = '{:>{width}}'.format(lcontext[-width:], width=width)
            rdisplay = '{:{width}}'.format(rcontext[:width], width=width)
            print(ldisplay, rdisplay)

    def _stem(self, word):
        return self._stemmer.stem(word).lower()
```

# 3.7 - Regular Expressions for Tokenizing Text


## Simple Approaches to Tokenization


```python
raw = """'When I'M a Duchess,' she said to herself, (not in a very hopeful tone
though), 'I won't have any pepper in my kitchen AT ALL. Soup does very
well without--Maybe it's always pepper that makes people hot-tempered,'..."""

print(re.split(r' ', raw)[:16])
print(re.split(r'[ \t\n]+', raw)[:16])
print(re.split(r'\s+', raw)[:16])
```

`[ \t\n]+` khớp một hoặc nhiều space, tab hoặc newline. `\s+` tổng quát hơn vì `\s` bao gồm mọi whitespace thông dụng.



```python
# Split tại chuỗi ký tự không phải word-character.
split_tokens = re.split(r'\W+', raw)

# Findall lấy trực tiếp các chuỗi word-character.
found_tokens = re.findall(r'\w+', raw)

print(split_tokens[:18])
print(found_tokens[:18])
```

`\w` là word-character; trong Python 3 nó hỗ trợ ký tự Unicode theo mặc định. `\W` là phần bù của `\w`.

`re.split(r'\W+', raw)` tạo chuỗi rỗng ở đầu/cuối nếu separator xuất hiện ngay biên chuỗi. `re.findall(r'\w+', raw)` tránh chuỗi rỗng, nhưng làm mất dấu câu và tách contraction như `won't` thành `won`, `t`.


```python
simple_pattern = r'\w+|\S\w*'
print(re.findall(simple_pattern, raw)[:30])
```

```python
better_pattern = r"\w+(?:[-']\w+)*|'|[-.(]+|\S\w*"
better_tokens = re.findall(better_pattern, raw)
print(better_tokens)
```

Phân tích pattern:

| Thành phần | Ý nghĩa |
| --- | --- |
| `\w+(?:[-']\w+)*` | từ có thể chứa hyphen/apostrophe bên trong: `hot-tempered`, `it's` |
| `'` | dấu nháy đơn đứng riêng |
| `[-.(]+` | một hay nhiều dấu `-`, `.`, `(`; giữ `--` và `...` thành token |
| `\S\w*` | trường hợp dự phòng: ký tự không phải whitespace và phần word theo sau |
| `|` | thử các lựa chọn từ trái sang phải |



## Regular Expression Character Classes

| Regex | Ý nghĩa |
| --- | --- |
| `\b` | biên từ, độ rộng bằng 0 |
| `\d` / `\D` | chữ số / không phải chữ số |
| `\s` / `\S` | whitespace / không phải whitespace |
| `\w` / `\W` | word-character / không phải word-character |
| `\t` | tab |

## NLTK's Regular Expression Tokenizer

`nltk.regexp_tokenize()` chuyên cho tokenization bằng regex. Flag `(?x)` bật **verbose mode**, cho phép xuống dòng và ghi comment để pattern dễ đọc.


```python
text = 'That U.S.A. poster-print costs $12.40 and 82%...'
pattern = r'''(?x)
    (?:[A-Z]\.)+        # abbreviation: U.S.A.
  | \$?\d+(?:\.\d+)?%? # tiền, số thập phân, phần trăm
  | \w+(?:-\w+)*        # word, có thể có hyphen bên trong
  | \.\.\.              # ellipsis
  | [][.,;"'?():_`-]    # dấu câu đứng riêng
'''

tokens_37 = nltk.regexp_tokenize(text, pattern)
print(tokens_37)
```

Trong verbose mode, whitespace không được xem là ký tự cần match. Muốn match space thật, dùng `\s` hoặc escape phù hợp.

Nếu `gaps=True`, pattern mô tả **khoảng trống giữa token**, tương tự `re.split()`.


```python
text = "Tokenize\tthis text\nby whitespace."
print(nltk.regexp_tokenize(text, r'\s+', gaps=True))
```

## Word Segmentation


Biểu diễn đường biên bằng một bit-string:

- `1`: có biên sau ký tự hiện tại.
- `0`: không có biên.
- Với text dài `n`, chỉ có `n - 1` vị trí có thể đặt biên.


```python
def segment(text, segs):
  words = []
  last = 0
  for i in range(len(segs)):
    if segs[i] == '1':
      words.append(text[last:i + 1])
      last = i + 1
  words.append(text[last:])
  return words

text = "doyouseethekittyseethedoggydoyoulikethekittylikethedoggy"
seg1 = "0000000000000001000000000010000000000000000100000000000"
seg2 = "0100100100100001001001000010100100010010000100010010000"

print(segment(text, seg1))
print(segment(text, seg2))
```

## Objective Function

Hàm mục tiêu để chấm điểm một phương án segmentation.

- `text_size`: số token cần dùng để tái tạo text.
- `lexicon_size`: tổng số ký tự của các từ duy nhất, cộng một ký tự phân cách cho mỗi từ.
- Điểm càng nhỏ càng tốt: mô hình ưu tiên cả cách giải thích ngắn và lexicon gọn.


```python
def evaluate(text, segs):
  words = segment(text, segs)
  text_size = len(words)
  lexicon_size = sum(len(word) + 1 for word in set(words))
  return text_size + lexicon_size

seg3 = "0000100100000011001000000110000100010000001100010000001"

print(segment(text, seg3))
print('seg1:', evaluate(text, seg1))
print('seg2:', evaluate(text, seg2))
print('seg3:', evaluate(text, seg3))
```

## Non-Deterministic Search with Simulated Annealing

Không thể thử mọi bit-string khi text dài vì có `2^(n-1)` phương án. **Simulated annealing** tìm kiếm gần đúng:

1. Bắt đầu từ một segmentation.
2. Lật ngẫu nhiên một số bit để tạo nghiệm lân cận.
3. Giữ nghiệm có điểm tốt hơn.
4. Hạ “nhiệt độ” dần để số bit bị xáo trộn ngày càng ít.


```python
from random import Random

rng = Random(7)  # seed cố định để ví dụ có thể tái lập

def flip(segs, pos):
  return segs[:pos] + str(1 - int(segs[pos])) + segs[pos + 1:]

def flip_n(segs, n, rng):
  for _ in range(n):
    segs = flip(segs, rng.randint(0, len(segs) - 1))
  return segs

def anneal(text, segs, iterations, cooling_rate, rng):
  temperature = float(len(segs))
  history = []
  while temperature > 0.5:
    best_segs, best = segs, evaluate(text, segs)
    for _ in range(iterations):
      guess = flip_n(segs, round(temperature), rng)
      score = evaluate(text, guess)
      if score < best:
        best, best_segs = score, guess
    segs = best_segs
    history.append((best, segment(text, segs)))
    temperature /= cooling_rate
  return segs, history

best_segs, history = anneal(text, seg1, iterations=300, cooling_rate=1.5, rng=rng)
print('Điểm cuối:', evaluate(text, best_segs))
print(segment(text, best_segs))
```

## Lining Things Up

Cú pháp tổng quát là `{value:format_spec}`:

| Ví dụ | Ý nghĩa |
| --- | --- |
| `{:6}` | độ rộng tối thiểu 6 |
| `{:<6}` | căn trái |
| `{:>6}` | căn phải |
| `{:^6}` | căn giữa |
| `{:.4f}` | số thực với 4 chữ số sau dấu thập phân |
| `{:.2%}` | hiển thị phần trăm với 2 chữ số thập phân |


```python
import math

print('|{:6}|'.format(41))
print('|{:<6}|'.format(41))
print('|{:6}|'.format('dog'))
print('|{:>6}|'.format('dog'))
print('{:.4f}'.format(math.pi))

count, total = 3205, 9375
print('accuracy for {} words: {:.4%}'.format(total, count / total))
```

## Tabulating Results

Tách riêng bước **xử lý ngôn ngữ** (tạo distribution) khỏi bước **trình bày** (in bảng) giúp code dễ kiểm tra và tái sử dụng.


```python
def tabulate(cfdist, words, categories):
  width = max(16, max(len(category) for category in categories) + 1)
  print('{:{width}}'.format('Category', width=width), end=' ')
  for word in words:
    print('{:>6}'.format(word), end=' ')
  print()

  for category in categories:
    print('{:{width}}'.format(category, width=width), end=' ')
    for word in words:
      print('{:6}'.format(cfdist[category][word]), end=' ')
    print()

nltk.download('brown')
from nltk.corpus import brown

cfd = nltk.ConditionalFreqDist(
  (genre, word)
  for genre in brown.categories()
  for word in brown.words(categories=genre)
)

genres = ['news', 'religion', 'hobbies', 'science_fiction', 'romance', 'humor']
modals = ['can', 'could', 'may', 'might', 'must', 'will']
tabulate(cfd, modals, genres)
```

## Text Wrapping



```python
from textwrap import fill

saying = ['After', 'all', 'is', 'said', 'and', 'done', ',',
          'more', 'is', 'said', 'than', 'done', '.']

pieces = ['{} ({})'.format(word, len(word)) for word in saying]
output = ', '.join(pieces)
wrapped = fill(output, width=65)
print(wrapped)
```
